```python
# 28. Create a DataFrame named school_df from a dictionary containing Student, Class, and Section. Construct a second DataFrame requesting the columns Student, Class, Section, and House, even though House is not present in the original data. Then assign one common value to the complete House column.
school_df = pd.DataFrame({
    'Student': ['A', 'B', 'C', 'D', 'E'],
    'Class': ['10th', '11th', '12th', '9th', '10th'],
    'Section': ['A', 'B', 'C', 'A', 'B']
})
print(school_df)
new_school_df = school_df.reindex(columns=['Student', 'Class', 'Section', 'House'])
new_school_df['House'] = 'Gryffindor'
print(new_school_df)
```
